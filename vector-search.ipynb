```python
books["tagged_description"].to_csv("tagged_description.txt",
                                   index = False,
                                   header = False,
                                   lineterminator="\n")
```

```python
raw_documents = TextLoader("tagged_description.txt", encoding="utf-8").load()
text_splitter = CharacterTextSplitter(chunk_size=1, chunk_overlap=0, separator="\n")
documents = text_splitter.split_documents(raw_documents)
```

```python
documents[1]
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
db_books = Chroma.from_documents(
    documents,
    embedding=embedding_model,
    persist_directory="./chroma.db"
)
```

```python
query = "A story about friendship, magic, and adventure"

results = db_books.similarity_search(
    query,
    k=5
)

for i, doc in enumerate(results, start=1):
    print(f"\nResult {i}")
    print(doc.page_content[:300])
    print("-" * 50)
```

```python
def recommend_books(query, k=5, min_rating=None):
    # 1. Search for more candidates than we need
    candidate_results = db_books.similarity_search(
        query,
        k=max(k * 5, k)
    )

    # 2. Extract ISBNs in relevance order
    isbn_list = [
        doc.page_content.split()[0].strip('"')
        for doc in candidate_results
    ]

    # 3. Prepare the book metadata
    books_lookup = books.copy()
    books_lookup["isbn13"] = books_lookup["isbn13"].astype(str)
    books_lookup = books_lookup.set_index("isbn13")

    # 4. Retrieve books in ChromaDB's search order
    recommended_books = books_lookup.reindex(isbn_list)

    # 5. Remove books that were not found
    recommended_books = recommended_books.dropna(how="all")

    # 6. Apply the minimum rating filter if requested
    if min_rating is not None:
        recommended_books = recommended_books[
            recommended_books["average_rating"] >= min_rating
        ]

    # 7. Keep only the requested number of recommendations
    recommended_books = recommended_books.head(k)

    # 8. Select useful information
    columns_to_display = [
        "title",
        "authors",
        "categories",
        "description",
        "thumbnail",
        "average_rating"
    ]

    recommended_books = recommended_books[columns_to_display]

    # 9. Return the recommendations
    return recommended_books.reset_index()
```

```python
recommend_books(
    "A fantasy story about magic and friendship",
    k=5
)
```

```python
recommend_books("A political book about governance",
                k=10,
                min_rating=3.9)
```
